기존 검증 구간은 반복 사용된 개발·재확인 구간으로 해석합니다.

```python
from pathlib import Path
import json
import pandas as pd

# Read the saved results only; do not retrain or evaluate the test set.
ROOT = Path.cwd()
APP = ROOT if (ROOT / "beer_artifacts").exists() else ROOT / "quietrisk"
ART = APP / "beer_artifacts"
metrics = json.loads((ART / "metrics.json").read_text(encoding="utf-8"))
monthly_rf = pd.read_csv(ART / "validation_monthly_RF.csv")
monthly_rule = pd.read_csv(ART / "validation_monthly_Recency.csv")
selected_rf = pd.read_csv(ART / "validation_selected_RF.csv.gz",
                          dtype={"Customer_ID": str})
selected_rule = pd.read_csv(ART / "validation_selected_Recency.csv.gz",
                            dtype={"Customer_ID": str})

model_hits = set(selected_rf.loc[selected_rf.y_D180.eq(1), "Customer_ID"])
rule_hits = set(selected_rule.loc[selected_rule.y_D180.eq(1), "Customer_ID"])
print("RF unique hits:", len(model_hits))
print("Recency unique hits:", len(rule_hits))
print("Net unique gain:", len(model_hits) - len(rule_hits))
print("Model only / Rule only / Both:",
      len(model_hits - rule_hits), len(rule_hits - model_hits),
      len(model_hits & rule_hits))
monthly_rf[["prediction_date", "selected", "hits", "recency_hits", "delta"]]
```
